```python
import requests
import pandas as pd
import json
import time

# API Configuration
API_KEY = "ehn7ry9bqmfJP5hm6byKicFijR93JAOr8V73cBaz"
BASE_URL = "https://api.sportradar.com/ncaafb/trial/v7/en"
HEADERS = {
    "accept": "application/json",
    "x-api-key": API_KEY
}

# ENDPOINT 1: Rankings by Poll (Alternative format)
def get_rankings_by_poll_df(poll_name, year, month):

    url = f"{BASE_URL}/polls/{poll_name}/{year}/{month}/rankings.json"

    print(f"Fetching rankings for {poll_name} - {year}/{month}...")
    print(f"URL: {url}")

    response = requests.get(url, headers=HEADERS)
    time.sleep(1)

    print(f"Status Code: {response.status_code}")

    if response.status_code == 200:
        rdata = response.json()

        # Extract poll information
        poll_id = rdata.get("poll", {}).get("id")
        poll_name = rdata.get("poll", {}).get("name")
        poll_alias = rdata.get("poll", {}).get("alias")
        season = rdata.get("season")
        week = rdata.get("week")
        effective_time = rdata.get("effective_time") # Corrected from effective_date

        rankings_info = []

        # Loop through rankings
        rankings = rdata.get("rankings", [])

        for r in rankings:
            rankings_details = {
                "ranking_id":r.get("id"), # Corrected to get 'id' from 'r'
                "poll_id": poll_id,
                "poll_name": poll_name,
                "season_id": season,
                "week": week,
                "effective_time": effective_time,
                "team_id": r.get("id"), # Corrected to get 'id' from 'r'
                "rank": r.get("rank"),
                "previous_rank": r.get("previous_rank"),
                "points": r.get("points"),
                "fp_votes": r.get("fp_votes"),
                "wins": r.get("wins"),
                "losses":r.get("losses"),
                "ties":r.get("ties")
            }
            rankings_info.append(rankings_details)

        ranking_df = pd.DataFrame(rankings_info)
        print(f"✓ Created DataFrame with {len(ranking_df)} rows")
        return ranking_df
    else:
        print(f"✗ Error {response.status_code}")
        print(f"Response: {response.text}")
        return pd.DataFrame()

# Usage
rankings_df = get_rankings_by_poll_df("AP25", "2025", "01")
print(rankings_df.head())
```

```python
import pandas as pd
import requests
import time
from datetime import datetime, timezone
import os

# ===========================
# CONFIG
# ===========================

API_KEY = "5P1abAwBsIFKBjtk3NvJGrABAv0vWWf99uZahH6Z"
BASE_URL = "https://api.sportradar.com/ncaafb/trial/v7/en"

BATCH_SIZE = 20
DELAY_SECONDS = 1.5

MAX_PLAYERS = 500   # ⭐ FINAL DECISION

PLAYERS_FILE = "/content/players.csv"
OUTPUT_FILE = "/content/player_statistics.csv"

# ===========================
# LOAD PLAYERS
# ===========================

players_df = pd.read_csv(PLAYERS_FILE)

# LIMIT TO 500 PLAYERS
players_df = players_df.head(MAX_PLAYERS)

total_players = len(players_df)
print(f"Processing {total_players} players (limited dataset)")

# ===========================
# RESUME SUPPORT
# ===========================

if os.path.exists(OUTPUT_FILE):
    existing_df = pd.read_csv(OUTPUT_FILE)
    fetched_players = set(existing_df["player_id"].astype(str).unique())
    print(f"Resuming... Already fetched {len(fetched_players)} players")
else:
    existing_df = pd.DataFrame()
    fetched_players = set()

# ===========================
# STORAGE
# ===========================

player_stats_list = []
stop_now = False

# ===========================
# FETCH LOOP
# ===========================

for i in range(0, total_players, BATCH_SIZE):
    batch = players_df.iloc[i:i + BATCH_SIZE]
    print(f"\nProcessing players {i+1} to {i+len(batch)}")

    for _, row in batch.iterrows():
        player_id = str(row["player_id"])

        if player_id in fetched_players:
            continue

        player_name = row.get("full_name")
        if pd.isna(player_name) or str(player_name).strip() == "":
            player_name = f"Player_{player_id[:6]}"

        url = f"{BASE_URL}/players/{player_id}/profile.json?api_key={API_KEY}"

        try:
            response = requests.get(url, timeout=30)
            time.sleep(DELAY_SECONDS)

            # ===========================
            # STOP CONDITIONS
            # ===========================

            if response.status_code == 429:
                print("\n🚫 API RATE LIMIT REACHED (429)")
                print("👉 Wait a few minutes and run again")
                stop_now = True
                break

            if response.status_code == 403:
                print(f"\n🚫 403 Forbidden for {player_name}")
                print("👉 Endpoint may not be available in your plan")
                stop_now = True
                break

            if response.status_code != 200:
                print(f"❌ Failed: {player_name} → {response.status_code}")
                continue

            data = response.json()

            # ================= PLAYER STATS =================
            for season in data.get("seasons", []):
                stats = season.get("statistics", {})

                player_stats_list.append({
                    "player_id": player_id,
                    "player_name": player_name,
                    "team_id": season.get("team", {}).get("id"),
                    "team_name": season.get("team", {}).get("name"),
                    "season_id": season.get("id"),
                    "season_year": season.get("year"),
                    "season_type": season.get("type"),
                    "games_played": stats.get("games_played"),
                    "games_started": stats.get("games_started"),
                    "rushing_yards": stats.get("rushing", {}).get("yards"),
                    "rushing_touchdowns": stats.get("rushing", {}).get("touchdowns"),
                    "receiving_yards": stats.get("receiving", {}).get("yards"),
                    "receiving_touchdowns": stats.get("receiving", {}).get("touchdowns"),
                    "kick_return_yards": stats.get("kick_returns", {}).get("yards"),
                    "fumbles": stats.get("fumbles"),
                    "fetched_at": datetime.now(timezone.utc)
                })

            print(f"✅ {player_name}")

        except Exception as e:
            print(f"⚠ Error fetching {player_name}: {e}")

    # ===========================
    # SAVE PROGRESS
    # ===========================

    if player_stats_list:
        batch_df = pd.DataFrame(player_stats_list)

        if not existing_df.empty:
            final_df = pd.concat([existing_df, batch_df], ignore_index=True)
        else:
            final_df = batch_df

        final_df.to_csv(OUTPUT_FILE, index=False)
        print("💾 Progress saved to player_statistics.csv")

        existing_df = final_df
        fetched_players = set(existing_df["player_id"].astype(str).unique())
        player_stats_list.clear()

    if stop_now:
        break

# ===========================
# FINAL SAVE
# ===========================

if player_stats_list:
    final_df = pd.concat([existing_df, pd.DataFrame(player_stats_list)], ignore_index=True)
    final_df.to_csv(OUTPUT_FILE, index=False)

print("\n🎉 Player statistics fetch completed!")
```
